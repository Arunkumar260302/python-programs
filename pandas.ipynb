```python
movies_data['release_year'].value_counts().head(10)
```

```python
import matplotlib as plt
top_years=movies_data['release_year'].value_counts().head(10)
top_years.plot(kind='bar')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
movies_data['type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('movies vs TV shows')
plt.ylabel('')
plt.show()
```

```python
movies_data['country'].value_counts().head(10)
```

```python
top_countries=movies_data['country'].value_counts().head(10)
top_countries.plot(kind='bar')
plt.title('Top 10 countries by number of titles')
plt.xlabel('Country')
plt.ylabel('Number of titles')
plt.show()
```

```python
type_percentage=movies_data['type'].value_counts(normalize=True)*100
type_percentage
```

```python
movies_data['rating'].value_counts().head(10)
```

```python
Top_ratings=movies_data['rating'].value_counts().head(10)
Top_ratings.plot(kind='bar')
plt.title('Top 10 Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of titles')
plt.show()
```

```python
yearly_titles=movies_data['release_year'].value_counts().sort_index()
yearly_titles.plot(kind='line')
plt.title('Titles Released over the years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

```python
year_type=movies_data.pivot_table(
    index='release_year',
    columns='type',
    aggfunc='size',
    fill_value=0
)
year_type.plot(kind='line')
plt.title('Movies vs TV shows Over the years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```
